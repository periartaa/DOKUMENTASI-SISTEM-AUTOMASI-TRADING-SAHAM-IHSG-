```python
# ==============================================================================
# 4. TAHAP 1: BULK SCREENING ANOMALI
# ==============================================================================
print("Sedang melakukan screening anomali saham...")
data = yf.download(tickers, start=start_date, end=end_date, group_by='ticker', auto_adjust=False)

saham_lolos = []

for ticker in tickers:
    try:
        df_single = data if len(tickers) == 1 else data[ticker].dropna()
        if len(df_single) < 21:
            continue

        last_row = df_single.iloc[-1]
        vol_sma20 = df_single['Volume'].iloc[-21:-1].mean()
        rvol = last_row['Volume'] / vol_sma20 if vol_sma20 > 0 else 0
        turnover_val = last_row['Volume'] * last_row['Close']
        price_change_pct = ((last_row['Close'] - last_row['Open']) / last_row['Open']) * 100

        # Syarat Anomali: RVOL >= 2.5x, Turnover >= Rp2 Miliar, Serba Positif (Close > Open & Change >= 2.5%)
        if (rvol >= 2.5) and (turnover_val >= 2_000_000_000) and (last_row['Close'] > last_row['Open']) and (price_change_pct >= 2.5):
            saham_lolos.append(ticker)
    except Exception:
        continue

print(f"-> Selesai! Saham kandidat yang lolos screening: {saham_lolos}\n")
```

```python
# ==============================================================================
# 5. TAHAP 2: DEEP ANALYSIS & TRADING PLAN
# ==============================================================================
if saham_lolos:
    hari_dict = {
        'Monday': 'Senin', 'Tuesday': 'Selasa', 'Wednesday': 'Rabu',
        'Thursday': 'Kamis', 'Friday': 'Jumat', 'Saturday': 'Sabtu', 'Sunday': 'Minggu'
    }

    for ticker in saham_lolos:
        df_single = data if len(tickers) == 1 else data[ticker].dropna()
        if isinstance(df_single.columns, pd.MultiIndex):
            df_single.columns = df_single.columns.get_level_values(0)

        df_filtered = df_single[df_single.index <= tanggal_target]
        if df_filtered.empty:
            print(f"Data untuk {ticker} tidak ditemukan pada/sebelum tanggal {tanggal_manual_str}.\n")
            continue

        last_date = df_filtered.index[-1]
        last_row = df_filtered.iloc[-1]
        close_today = last_row['Close']

        next_bday = last_date + pd.tseries.offsets.BDay(1)

        sl_price = close_today * 0.97
        trailing_trigger = close_today * 1.07
        trailing_lock = close_today * 1.05

        print(f"==================================================")
        print(f" TRADING PLAN: {ticker}")
        print(f"==================================================")
        print(f". Tanggal Sinyal (YYYY-MM-DD): {last_date.strftime('%Y-%m-%d')} ({hari_dict[last_date.day_name()]})")
        print(f". Target Eksekusi (YYYY-MM-DD): {next_bday.strftime('%Y-%m-%d')} ({hari_dict[next_bday.day_name()]})")
        print(f"--------------------------------------------------")
        print(f". Harga Beli Acuan : Rp{close_today:,.0f}")
        print(f". Cut Loss (-3%)    : Rp{sl_price:,.0f}")
        print(f". Trailing Trigger  : Rp{trailing_trigger:,.0f} (+7%)")
        print(f". Trailing Lock     : Rp{trailing_lock:,.0f} (+5%)")
        print(f". Batas Hold Max    : 5 Hari Bursa")
        print(f"==================================================\n")
else:
    print("Hasil: Tidak ada saham yang memicu sinyal beli hari ini.")
```
